### Rule

I will prioritize content that is both stale and has meaningful search volume.

The rule uses two observable signals:
- `days_since_last_update`: higher values indicate more stale content and support a refresh action.
- `search_volume`: higher values indicate greater search demand and support prioritizing the item.

The baseline score will give more points to content that is older and has higher search volume. Higher scores mean higher priority for review/action.

### Reason codes

- `STALE_HIGH_VOLUME`: the content is relatively stale and has relatively high search volume.
- `STALE`: the content is relatively stale but does not have high search volume.
- `HIGH_VOLUME`: the content has relatively high search volume but is not especially stale.
- `LOW_PRIORITY`: neither signal is high enough to justify a strong action.

### Action labels

- `REFRESH`: stale and high-volume content.
- `REVIEW`: one of the two signals is high.
- `MONITOR`: neither signal is high.

```python
# Section 1: signal checks and rule setup

import pandas as pd
import numpy as np

# Work from the already-loaded dataset
work = df.copy()

# ---------------------------------------------------------
# Signal 1: days_since_last_update
# ---------------------------------------------------------

stale = work["days_since_last_update"].dropna()

stale_bins = [-np.inf, 30, 90, 180, np.inf]
stale_labels = ["0-30", "31-90", "91-180", "181+"]

work["staleness_bucket"] = pd.cut(
    work["days_since_last_update"],
    bins=stale_bins,
    labels=stale_labels
)

stale_check = (
    work.groupby("staleness_bucket", observed=False)
        .agg(
            n=("content_id", "size"),
            median_trend_pct=("trend_pct", "median"),
            mean_trend_pct=("trend_pct", "mean")
        )
        .reset_index()
)

print("SIGNAL 1 — days_since_last_update")
display(stale_check)

# ---------------------------------------------------------
# Signal 2: search_volume
# ---------------------------------------------------------

volume = work["search_volume"].dropna()

volume_bins = [-np.inf, 10, 100, 1000, np.inf]
volume_labels = ["0-10", "11-100", "101-1000", "1001+"]

work["volume_bucket"] = pd.cut(
    work["search_volume"],
    bins=volume_bins,
    labels=volume_labels
)

volume_check = (
    work.groupby("volume_bucket", observed=False)
        .agg(
            n=("content_id", "size"),
            median_trend_pct=("trend_pct", "median"),
            mean_trend_pct=("trend_pct", "mean")
        )
        .reset_index()
)

print("SIGNAL 2 — search_volume")
display(volume_check)

# ---------------------------------------------------------
# Verdicts
# ---------------------------------------------------------

print("VERDICTS")
print("days_since_last_update: CONFIRMED if the more-stale buckets show a meaningfully different/directionally worse trend than the freshest bucket.")
print("search_volume: CONFIRMED if higher-volume buckets show a meaningful directional difference in trend or opportunity.")
```

### Signal verdicts

- **days_since_last_update — CONFIRMED:** The measured buckets show a directional relationship: the freshest bucket had mean trend_pct of +0.78%, while older buckets were negative, including -15.68% for 91–180 days. The 181+ bucket was also negative at -6.78%. This supports using staleness as a refresh-priority signal, while recognizing that the relationship is not perfectly monotonic.

- **search_volume — CONFIRMED:** Higher search-volume buckets showed a stronger directional opportunity. Mean trend_pct was -9.81% for 0–10 volume, +1.74% for 101–1000, and +20.23% for 1001+. This supports using search volume as a prioritization signal.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# Section 2: Build the ranked baseline action queue

# Start from the working dataframe
queue = work.copy()

# ---------------------------------------------------------
# 1. Create simple, observable signal scores
# ---------------------------------------------------------

# Staleness score
queue["stale_score"] = pd.cut(
    queue["days_since_last_update"],
    bins=[-np.inf, 30, 90, 180, np.inf],
    labels=[0, 1, 2, 3]
).astype(float).fillna(0)

# Search-volume score
queue["volume_score"] = pd.cut(
    queue["search_volume"],
    bins=[-np.inf, 10, 100, 1000, np.inf],
    labels=[0, 1, 2, 3]
).astype(float).fillna(0)

# ---------------------------------------------------------
# 2. Combined baseline score
# ---------------------------------------------------------

queue["score"] = queue["stale_score"] + queue["volume_score"]

# ---------------------------------------------------------
# 3. ONE reason code per row
# ---------------------------------------------------------

queue["reason_code"] = np.select(
    [
        (queue["stale_score"] >= 2) & (queue["volume_score"] >= 2),
        (queue["stale_score"] >= 2),
        (queue["volume_score"] >= 2)
    ],
    [
        "STALE_HIGH_VOLUME",
        "STALE",
        "HIGH_VOLUME"
    ],
    default="LOW_PRIORITY"
)

# ---------------------------------------------------------
# 4. Action label
# ---------------------------------------------------------

queue["action"] = np.select(
    [
        queue["reason_code"] == "STALE_HIGH_VOLUME",
        queue["reason_code"].isin(["STALE", "HIGH_VOLUME"])
    ],
    [
        "REFRESH",
        "REVIEW"
    ],
    default="MONITOR"
)

# ---------------------------------------------------------
# 5. Rank the queue
# ---------------------------------------------------------

queue = queue.sort_values(
    ["score", "search_volume", "days_since_last_update"],
    ascending=[False, False, False]
).reset_index(drop=True)

queue["rank"] = queue.index + 1

# ---------------------------------------------------------
# 6. Select output columns
# ---------------------------------------------------------

output_cols = [
    "rank",
    "content_id",
    "score",
    "reason_code",
    "action",
    "days_since_last_update",
    "search_volume"
]

baseline_queue = queue[output_cols].copy()

# ---------------------------------------------------------
# 7. Write the required CSV
# ---------------------------------------------------------

output_path = "work/outputs/baseline_action_score.csv"

baseline_queue.to_csv(output_path, index=False)

print("CSV written to:", output_path)
print("Rows:", len(baseline_queue))

print("\nTop 10:")
display(baseline_queue.head(10))
```

```python
# Section 3: Top-20 review

top20 = baseline_queue.head(20).copy()

# Add simple confidence notes based on the baseline score
top20["confidence_note"] = np.select(
    [
        top20["score"] >= 5,
        top20["score"] >= 4,
        top20["score"] >= 2
    ],
    [
        "High: both signals are strongly elevated.",
        "Medium-high: both signals contribute meaningfully.",
        "Medium: one signal is elevated."
    ],
    default="Low: neither signal is strongly elevated."
)

# Add a skeptical "what would make it wrong" note
top20["what_would_make_it_wrong"] = np.select(
    [
        top20["reason_code"] == "STALE_HIGH_VOLUME",
        top20["reason_code"] == "STALE",
        top20["reason_code"] == "HIGH_VOLUME"
    ],
    [
        "It would be wrong if the page is intentionally old or already scheduled for an update.",
        "It would be wrong if the content is still performing well despite being old.",
        "It would be wrong if high search volume does not translate into a meaningful opportunity for this content.",
    ],
    default="It would be wrong if there is no meaningful reason to prioritize the content."
)

# Display the required top-20 review
review_cols = [
    "rank",
    "content_id",
    "action",
    "reason_code",
    "confidence_note",
    "what_would_make_it_wrong"
]

top20_review = top20[review_cols].copy()

display(top20_review)
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
# Section 4: Weak picks + leakage check

# ---------------------------------------------------------
# 1. Identify weak/questionable picks
# ---------------------------------------------------------

weak_picks = baseline_queue[
    (baseline_queue["score"] <= 2)
].head(10).copy()

print("WEAK / QUESTIONABLE PICKS")
display(weak_picks)

# ---------------------------------------------------------
# 2. Leakage check
# ---------------------------------------------------------

rule_inputs = [
    "days_since_last_update",
    "search_volume"
]

print("\nRULE INPUTS:")
print(rule_inputs)

# Check that the scoring rule only depends on the two chosen signals
print("\nLEAKAGE CHECK")
print("The baseline score uses only:")
print("- days_since_last_update")
print("- search_volume")

print("\nNo future-window or label-derived fields are used in the score.")

# Explicitly check that future/label-like fields are not used
forbidden_terms = [
    "label",
    "target",
    "outcome",
    "future",
    "next",
    "conversion"
]

used_columns = set(["days_since_last_update", "search_volume"])

possible_leakage = [
    col for col in used_columns
    if any(term in col.lower() for term in forbidden_terms)
]

print("\nPossible leakage fields among rule inputs:", possible_leakage)

if len(possible_leakage) == 0:
    print("LEAKAGE CHECK: PASS")
else:
    print("LEAKAGE CHECK: REVIEW NEEDED")

# ---------------------------------------------------------
# 3. Weak-pick explanation
# ---------------------------------------------------------

print("\nWEAK PICK REVIEW")
print(
    "Weak picks are lower-scoring rows that may not deserve immediate action. "
    "They are useful as a skeptical check because the baseline is intentionally simple."
)
```

### Weak picks

The weakest picks are useful as a skeptical check. A low baseline score does not prove that an item has no opportunity; it only means that the two selected signals did not strongly prioritize it.

### Leakage check

The baseline score uses only `days_since_last_update` and `search_volume`. These are observable signals available for the rule. No future-window field, target/label, or outcome-derived field is used in the scoring calculation.
