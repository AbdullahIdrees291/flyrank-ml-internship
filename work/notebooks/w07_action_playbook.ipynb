## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
# ML-10 — Section 1
# Ranked actions + reason codes

import pandas as pd

# Example action queue structure
# Replace these rows later with your validated model output.

actions = pd.DataFrame({
    "priority": [1, 2, 3, 4, 5],
    "action": [
        "Refresh underperforming content",
        "Review high-potential content",
        "Improve content relevance",
        "Test a new content angle",
        "Leave stable content unchanged"
    ],
    "reason_code": [
        "LOW_PERFORMANCE",
        "HIGH_POTENTIAL",
        "RELEVANCE_OPPORTUNITY",
        "TEST_OPPORTUNITY",
        "STABLE"
    ],
    "recommended_action": [
        "Human review before refresh",
        "Prioritize for human review",
        "Review topic and audience fit",
        "Run a small controlled test",
        "Monitor without immediate change"
    ]
})

# Display ranked queue
print("Ranked Action Queue")
display(actions)

# Check reason codes
print("\nReason Code Summary")
display(actions["reason_code"].value_counts())

# Basic validation checks
assert actions["priority"].is_unique
assert actions["reason_code"].notna().all()
assert actions["recommended_action"].notna().all()

print("\nSection 1 checks passed.")
```

### Ranked actions + reason codes

The action queue prioritizes five practical content actions. The first priority is to review underperforming content before making a refresh decision. High-potential content is reviewed next, followed by opportunities to improve topic and audience relevance. New content angles should be tested on a small scale before wider use. Stable content should be monitored rather than changed without evidence.

Reason codes are used to keep the recommendations understandable: `LOW_PERFORMANCE`, `HIGH_POTENTIAL`, `RELEVANCE_OPPORTUNITY`, `TEST_OPPORTUNITY`, and `STABLE`.

These actions are decision-support recommendations, not automatic decisions. A human should review the evidence before taking action.


## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
# ML-10 — Section 2: Intended use and limits

print("Intended use:")
print("- Prioritize content actions for human review.")
print("- Support content refresh and testing decisions.")
print("- Use recommendations as decision-support, not automatic decisions.")

print("\nLimits:")
print("- Recommendations are directional.")
print("- They do not guarantee future performance.")
print("- Human review is required before action.")
print("- Recommendations may become stale when data or audience behavior changes.")

# Basic checks
intended_use = True
human_review_required = True
automatic_decisions = False

assert intended_use
assert human_review_required
assert automatic_decisions is False

print("\nSection 2 checks passed.")
```

### Intended use and limits

This playbook is intended to help content teams prioritize which content actions deserve human attention first. The recommendations are decision-support outputs based on the available model results and should be used to guide review, not to make automatic content decisions.

The playbook is useful for prioritizing content refreshes, reviewing promising content, identifying relevance opportunities, and selecting small tests. It is not intended to predict guaranteed future performance or replace human judgment.

The recommendations may become less reliable when audience behavior, content topics, platform conditions, or the underlying data change substantially. Results should therefore be treated as directional and reviewed against current evidence before action.


## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
# ML-10 — Section 3: Human review + no-go list

human_review_required = True

no_go_actions = [
    "automatic publishing",
    "automatic deletion",
    "automatic major content changes",
    "decisions about sensitive or high-impact content without human review"
]

print("Human review required:", human_review_required)

print("\nNo-go automation list:")
for item in no_go_actions:
    print("-", item)

# Basic checks
assert human_review_required is True
assert len(no_go_actions) > 0

print("\nSection 3 checks passed.")
```

### Human review + the no-go list

A person must review the evidence behind each recommendation before taking action. The reviewer should check the content context, audience fit, recent performance, and whether the recommendation still makes sense under current conditions.

The system should not automatically publish, delete, rewrite, or substantially change content. It should also not make decisions involving sensitive topics, individual people, or high-impact outcomes without appropriate human judgment.

The model output is a prioritization aid only. Final content decisions remain with a human reviewer.


## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

```python
# ML-10 — Section 4: Monitoring / retrain triggers

monitoring_triggers = [
    "Meaningful change in input data distribution",
    "Decline in recommendation quality",
    "Change in audience behavior",
    "Repeated human overrides",
    "Persistent performance degradation"
]

retrain_requires_human_review = True

print("Monitoring / retrain triggers:")
for trigger in monitoring_triggers:
    print("-", trigger)

print("\nRetraining requires human review:",
      retrain_requires_human_review)

# Basic checks
assert len(monitoring_triggers) > 0
assert retrain_requires_human_review is True

print("\nSection 4 checks passed.")
```

### Monitoring / retrain triggers

The recommendations should be monitored over time to check whether they remain useful. A refresh or retraining review should be considered when recent data shows a meaningful change from the data used to build or validate the model.

Useful warning signs include changes in input distributions, declining recommendation quality, changes in audience behavior, or repeated human overrides of the recommended actions.

Retraining should not happen automatically from a single unusual observation. A human should first confirm that the change is persistent and meaningful before deciding whether the model needs to be refreshed or retrained.
