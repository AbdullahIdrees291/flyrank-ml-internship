## 1. My lane as an ML task (type)

Task: Content Performance Decline Prediction

ML Task Type: Classification

I want to predict whether a content item is declining or not. This is a classification task because the outcome has two categories: declining and not declining.

```python
task_type = "classification"
print("ML task type:", task_type)
```

## 2. Target or proxy

Target: is_declining_label

I want to predict whether a content item is declining. The target is an observed outcome: it is 1 when trend_direction is "down" and 0 otherwise.

I will not use trend_direction as a model feature because it is used to create the target and would cause target leakage.

```python
import os

print(os.getcwd())
print(os.listdir())
```

```python
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print("Rows:", len(df))
print("Declining labels:", df["is_declining_label"].sum())
print("Declining rate:", round(df["is_declining_label"].mean() * 100, 1), "%")
```

## 3. Success metric

**Success metric: F1-score**

I will use F1-score to evaluate the classification model. It balances precision and recall, which is useful because we want to identify declining content while avoiding too many incorrect alerts.

```python
from sklearn.metrics import f1_score

print("Chosen success metric: F1-score")
```

## 4. The unit of analysis, as a real dataframe

**Unit of analysis: one row = one content item/page.**

Each row represents one content item. The dataframe shows content characteristics and performance-related measurements for each item.

```python
display_cols = [
    "content_id",
    "content_type",
    "word_count",
    "content_age_days",
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "engagement_rate",
    "is_declining_label"
]

df[display_cols].head(10)
```

## 5. Why ML beats a fixed rule here

A fixed rule such as "review every page with fewer than a certain number of impressions" would be too simple. Content performance can depend on several factors, including traffic, clicks, sessions, content age, content type, and engagement.

ML can combine multiple signals and learn patterns associated with declining content instead of relying on one manually chosen threshold.

The output would support a real content action: prioritizing pages for review or refresh. The model would provide decision-support rather than a guarantee.

```python
feature_candidates = [
    "word_count",
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "clicks_90d",
    "pageviews_90d",
    "sessions_90d",
    "users_90d",
    "engaged_sessions_90d",
    "engagement_rate",
    "scroll_rate",
    "avg_position"
]

available_features = [c for c in feature_candidates if c in df.columns]

print("Candidate feature count:", len(available_features))
print("Candidate features:")
print(available_features)
```
